```python
## 1. Initial Data Audit

The first step is to inspect the raw dataset before making
any modifications. This includes checking the dataset size,
column names, missing values, duplicate records, and data types.
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100
})

display(missing_report)
```

```python
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
df["Date"].head(20)
```

```python
#Data cleaning
#check original data type
df["Date"].dtype
```

```python
#convert date to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
#check data type after conversion
print(df["Date"].dtype)
```

```python
#check invalid /missing dates
print("Missing/Invalid dates:", df["Date"].isna().sum())
```

```python
#check date range
print("Earliest date:", df["Date"].min())
print("Latest date:", df["Date"].max())
```

```python
print(df["Date"].head(20))
```

```python
# Check missing values in every column

missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100
})

display(missing_report)
```

```python
#check duplicate rows
duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
```

```python
duplicate_order_ids = df["OrderID"].duplicated().sum()

print("Duplicate OrderIDs:", duplicate_order_ids)
```

```python
numeric_columns = [
    "Quantity",
    "UnitPrice",
    "ItemsInCart",
    "TotalPrice"
]

print(numeric_columns)
```

```python
df[numeric_columns].dtypes
```

```python
display(df[numeric_columns].describe())
```

```python
for column in numeric_columns:
    print(f"\n{column} - Negative values:")
    display(df[df[column] < 0])
```

```python
print(df[numeric_columns].dtypes)
```

```python
display(df[numeric_columns].describe())
```

```python
df["CalculatedTotal"] = df["Quantity"] * df["UnitPrice"]

df["PriceDifference"] = (
    df["TotalPrice"] - df["CalculatedTotal"]
)

print(
    "Mismatched TotalPrice:",
    (df["PriceDifference"].abs() > 0.01).sum()
)
```

```python
df.drop(
    columns=["CalculatedTotal", "PriceDifference"],
    inplace=True
)
```

```python
print(df.columns.tolist())
```

```python
text_columns = [
    "Product",
    "ShippingAddress",
    "PaymentMethod",
    "OrderStatus",
    "TrackingNumber",
    "CouponCode",
    "ReferralSource"
]

for column in text_columns:
    extra_spaces = (
        df[column].astype(str) !=
        df[column].astype(str).str.strip()
    ).sum()

    print(f"{column}: {extra_spaces} values with extra spaces")
```

```python
print(df["PaymentMethod"].value_counts())
print(df["OrderStatus"].value_counts())
print(df["ReferralSource"].value_counts())
for column in text_columns:
    extra_spaces = (
        df[column].astype(str) !=
        df[column].astype(str).str.strip()
    ).sum()

    print(f"{column}: {extra_spaces}")
```

```python
print("Unique OrderIDs:", df["OrderID"].nunique())
print("Unique CustomerIDs:", df["CustomerID"].nunique())
print("Unique TrackingNumbers:", df["TrackingNumber"].nunique())
```

```python
cleaning_summary = pd.DataFrame({
    "Check": [
        "Missing Values",
        "Duplicate Rows",
        "Duplicate OrderIDs",
        "Invalid Dates",
        "Negative Quantity",
        "Negative UnitPrice",
        "Negative ItemsInCart",
        "Negative TotalPrice",
        "TotalPrice Mismatch",
        "Extra Whitespace"
    ],
    "Result": [
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["OrderID"].duplicated().sum(),
        df["Date"].isna().sum(),
        (df["Quantity"] < 0).sum(),
        (df["UnitPrice"] < 0).sum(),
        (df["ItemsInCart"] < 0).sum(),
        (df["TotalPrice"] < 0).sum(),
        0,
        0
    ]
})

display(cleaning_summary)
```
