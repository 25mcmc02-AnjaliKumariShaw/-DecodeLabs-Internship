```python
Q1 = df["TotalPrice"].quantile(0.25)
Q3 = df["TotalPrice"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df["TotalPrice"] < lower_bound) |
    (df["TotalPrice"] > upper_bound)
]

print("Number of TotalPrice outliers:", len(outliers))

display(
    outliers[
        [
            "OrderID",
            "Product",
            "Quantity",
            "UnitPrice",
            "TotalPrice"
        ]
    ].sort_values(
        "TotalPrice",
        ascending=False
    ).head(20)
)
```
